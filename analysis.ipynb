```python
final_list = []  # initializing an empty list for the entire file

for line in open('billionaires.csv'):
    cleaned_line = line.strip()
    cleaned_list = cleaned_line.split(',')
    final_list.append(cleaned_list)  
printed_names = []
for row in final_list:
    name = row[0]
    country = row[10]
    if country == '"United States"' and name not in printed_names: #print only usa billionaires and not the same name of them 
        print(row)
        printed_names.append(name)
```

```python
gender_counts = billionaires['demographics_gender'].value_counts()
print(gender_counts)
```

```python
import matplotlib.pyplot as plt

gender_counts.plot(kind='bar')

plt.title('Number of Billionaires by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')

plt.show()
```

```python
import matplotlib.pyplot as plt

gender_counts.plot(kind='bar' ,color=['blue', 'pink'])

plt.title('Number of Billionaires by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')

plt.show()
```

```python
billionaires.groupby(['year', 'demographics_gender']).size().unstack().plot( #had to search (size = counts the rows) unstack = moves gender into columns and plot does the graphing
    kind='line',
    figsize=(12,6)
)
```

```python
billionaires.assign(
    wealth_simple= billionaires['wealth_type'].apply(
        lambda x: 'inherited' if str(x).lower() == 'inherited' else 'other'
    )
).groupby(['demographics_gender', 'wealth_simple']).size().unstack().plot(kind='bar')
```

```python
billionaires['country_simple'] = billionaires['location_country code'].apply(
    lambda x: 'USA' if str(x).upper() == 'USA' else 'Other'
)
billionaires['country_simple'].value_counts().plot(kind='bar', figsize=(6,4))

plt.title('Billionaires: USA vs Other Countries')
plt.xlabel('Country Group')
plt.ylabel('Number of People')

plt.show()
```

```python
billionaires.groupby('demographics_gender')['demographics_age'].mean().plot(kind='bar')
```

```python
female = billionaires[billionaires['demographics_gender'] == 'female']
female['wealth_how_industry'].value_counts().head(10).plot(kind='barh' , color='pink' )
plt.xlabel('Number of Female Billionaires')
plt.ylabel('Industry')
plt.title('Top 10 Industries of Female Billionaires')
```

```python
male = billionaires[billionaires['demographics_gender'] == 'male']

male['wealth_how_industry'].value_counts().head(10).plot(kind='barh', )

plt.xlabel('Number of Male Billionaires')
plt.ylabel('Industry')
plt.title('Top 10 Industries of Male Billionaires')

plt.show()
```

```python
billionaires.boxplot(
    column='wealth_worth in billions',
    by='demographics_gender',
    grid=False
)

plt.xlabel('Gender')
plt.ylabel('Net Worth (Billions USD)')
plt.title('Wealth Distribution: Male vs Female Billionaires')


plt.show()
```

```python
male = billionaires[billionaires['demographics_gender'] == 'male']
female = billionaires[billionaires['demographics_gender'] == 'female']

plt.scatter(
    male['demographics_age'],
    male['wealth_worth in billions'],
    alpha=0.5,
    label='Male'
)

plt.scatter(
    female['demographics_age'],
    female['wealth_worth in billions'],
    alpha=0.7,
    color='pink',
    label='Female'
)

plt.xlabel('Age')
plt.ylabel('Net Worth (Billions USD)')
plt.title('Age vs Net Worth of Billionaires by Gender')
plt.legend()

plt.show()
```

```python


top_countries = billionaires['location_citizenship'].value_counts().head(10)

top_countries.plot(
    kind='barh',
    figsize=(10, 6)
)
plt.xlabel('Number of Billionaires')
plt.ylabel('Country')
plt.title('Top 10 Countries with the Most Billionaires')

plt.show()
```

```python
top_countries = (
    billionaires['location_citizenship']
    .value_counts()
    .head(10)
    .sort_values(ascending=True)   # keeps biggest at top in barh
)

colors = [
    'red', 'orange', 'yellow', 'green', 'blue',
    'purple', 'pink', 'brown', 'gray', 'cyan'
]

top_countries.plot(
    kind='barh',
    figsize=(10, 6),
    color=colors
)

plt.xlabel('Number of Billionaires')
plt.ylabel('Country')
plt.title('Top 10 Countries with the Most Billionaires')
```

```python
female = billionaires[billionaires['demographics_gender'] == 'female']

top_female_countries = (
    female['location_citizenship']
    .value_counts()
    .head(10)
    .sort_values(ascending=True)
)

top_female_countries.plot(
    kind='barh',
    figsize=(10, 6),
    color='pink'
)

plt.xlabel('Number of Female Billionaires')
plt.ylabel('Country')
plt.title('Top 10 Countries with the Most Female Billionaires')

plt.show()
```

```python
male = billionaires[billionaires['demographics_gender'] == 'male']

top_male_countries = (
    male['location_citizenship']
    .value_counts()
    .head(10)
    .sort_values(ascending=True)
)

top_male_countries.plot(
    kind='barh',
    figsize=(10, 6),
    color='blue'
)

plt.xlabel('Number of Male Billionaires')
plt.ylabel('Country')
plt.title('Top 10 Countries with the Most Male Billionaires')

plt.show()
```

```python
gender_counts = billionaires['demographics_gender'].value_counts()

plt.pie(
    gender_counts,
    labels=gender_counts.index,
    autopct='%1.1f%%', #shows one decimal place in the displayed percentage
    colors=['lightblue', 'pink']
)

plt.title('Gender Distribution of Billionaires')
plt.show()
```

```python
billionaires[
    billionaires['demographics_gender'].str.lower().str.strip() != 'married couple'
].groupby(['year', 'demographics_gender']).size().unstack().plot(
    kind='line',
    figsize=(12,6),
    color=['blue', 'pink']
)
```

```python
ax = billionaires[
    billionaires['demographics_gender'].str.lower().str.strip() != 'married couple'
].groupby(['year', 'demographics_gender']).size().unstack().plot(
    kind='line',
    figsize=(12,6),
    color=['pink', 'blue']
)

ax.set_title("Billionaire Population Over Time by Gender")
ax.set_xlabel("Year")
ax.set_ylabel("Number of Billionaires")
```

```python
ax = billionaires.assign(
    wealth_simple=billionaires['wealth_type'].apply(
        lambda x: 'inherited' if str(x).lower() == 'inherited' else 'other'
    )
).groupby(['demographics_gender', 'wealth_simple']).size().unstack().plot(
    kind='bar',
    color=['red', 'yellow']
)

ax.set_title("Wealth Type Distribution by Gender")
ax.set_xlabel("Gender")
ax.set_ylabel("Number of Billionaires")
```
